```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Veriyi okuma
df = pd.read_excel('Başarı.xlsx', sheet_name='EVDS')
df['LOGINF'] = np.log(df['INF'])
df['LOGIMP'] = np.log(df['IMP'])
df['LOGEXC'] = np.log(df['EXC'])
df['LOGGDP'] = np.log(df['GDP'])

variables = ['LOGINF', 'LOGIMP', 'LOGEXC', 'LOGGDP']
trend_results = []

for var in variables:
    series = df[var].dropna()
    t = np.arange(len(series)) # Zaman indeksi (0, 1, 2, ...)
    X = sm.add_constant(t)     # Sabit terim ekle

    # Zaman trendine göre OLS regresyonu
    model = sm.OLS(series, X).fit()

    trend_results.append({
        'Değişken': var,
        'Trend Katsayısı': model.params[1],
        'Trend P-Değeri': model.pvalues[1],
        'Trend Durumu': 'Anlamlı Trend Var' if model.pvalues[1] < 0.05 else 'Trend Yok'
    })

trend_df = pd.DataFrame(trend_results)
pd.set_option('display.float_format', lambda x: '%.4f' % x)
display(trend_df)
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller

# Veriyi okuma ve log dönüşümleri
df = pd.read_excel('Başarı.xlsx', sheet_name='EVDS')
df['LOGINF'] = np.log(df['INF'])
df['LOGIMP'] = np.log(df['IMP'])
df['LOGEXC'] = np.log(df['EXC'])
df['LOGGDP'] = np.log(df['GDP'])

results_list = []
variables = ['LOGINF', 'LOGIMP', 'LOGEXC', 'LOGGDP']

for var in variables:
    # 1. Düzey için: Sabitli ve Trendli Model (regression='ct')
    res_lvl = adfuller(df[var].dropna(), regression='ct')
    results_list.append({
        'Değişken': var,
        'Model Türü': 'Sabitli ve Trendli (Level)',
        'ADF İstatistiği': res_lvl[0],
        'P-Değeri': res_lvl[1],
        '%1 Kritik Değer': res_lvl[4]['1%'],
        '%5 Kritik Değer': res_lvl[4]['5%'],
        '%10 Kritik Değer': res_lvl[4]['10%'],
        'Sonuç (p<0.05)': 'Durağan' if res_lvl[1] < 0.05 else 'Durağan Değil'
    })

    # 2. Birinci Fark için: Sabitli Model (regression='c')
    res_diff = adfuller(df[var].diff().dropna(), regression='c')
    results_list.append({
        'Değişken': f"Δ({var})",
        'Model Türü': 'Sabitli (1. Fark)',
        'ADF İstatistiği': res_diff[0],
        'P-Değeri': res_diff[1],
        '%1 Kritik Değer': res_diff[4]['1%'],
        '%5 Kritik Değer': res_diff[4]['5%'],
        '%10 Kritik Değer': res_diff[4]['10%'],
        'Sonuç (p<0.05)': 'Durağan' if res_diff[1] < 0.05 else 'Durağan Değil'
    })

res_df = pd.DataFrame(results_list)
pd.set_option('display.float_format', lambda x: '%.4f' % x)

print("=== SABİTLİ VE TRENDLİ / SABİTLİ ADF BİRİM KÖK TESTİ TABLOSU ===")
display(res_df)

# Markdown formatında çıktı almak isterseniz alttaki satırı açabilirsiniz:
# print(res_df.to_markdown(index=False))
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.ardl import ardl_select_order, ARDL

# 1. Veriyi okuma ve log dönüşümleri
df = pd.read_excel('Başarı.xlsx', sheet_name='EVDS')
df['LOGINF'] = np.log(df['INF'])
df['LOGIMP'] = np.log(df['IMP'])
df['LOGEXC'] = np.log(df['EXC'])
df['LOGGDP'] = np.log(df['GDP'])

# Bağımlı değişken (y) ve bağımsız değişkenler (X)
y = df['LOGGDP'].dropna()
X = df[['LOGINF', 'LOGIMP', 'LOGEXC']].loc[y.index]

# 2. Akaike (AIC) kriterine göre optimal gecikme uzunluğunu bulma
print("--- OPTİMAL GECİKME UZUNLUĞU SEÇİLİYOR ---")
selected_order = ardl_select_order(
    y,
    maxlag=4,
    exog=X,
    maxorder=4,
    ic='aic',
    trend='c'
)

print(f"Seçilen AR Gecikmesi (y): {selected_order.ar_lags}")
print(f"Seçilen DL Gecikmeleri (X): {selected_order.dl_lags}")

# Dışsal gecikmeleri sözlüğe çevirme
dl_orders = {k: max(v) for k, v in selected_order.dl_lags.items()}
ar_order = selected_order.ar_lags[0] if isinstance(selected_order.ar_lags, list) else selected_order.ar_lags

# 3. Modelin kurulması
ardl_model = ARDL(
    y,
    lags=ar_order,
    exog=X,
    order=dl_orders,
    trend='c'
).fit()

print("\n=== ARDL MODEL ÖZETİ ===")
print(ardl_model.summary())

# 4. Sınır Testi (Bounds Test) için Düzey Değişkenlerinin Wald Testi
# Modeldeki gecikmeli düzey (level) değişkenlerinin isimlerini bulma ve Wald testi yapma
level_params = [p for p in ardl_model.params.index if 'L1.' in p or 'level' in p.lower() or p in ['LOGGDP', 'LOGINF', 'LOGIMP', 'LOGEXC'] and 'l' not in p.lower()]
# Alternatif olarak statsmodels ardl yapısında düzey terimleri 'L1.LOGGDP' gibi geçer.
level_params_identified = [p for p in ardl_model.params.index if p.startswith('L1.')]

print("\n=== EŞBÜTÜNLEŞME (SINIR) WALD TESTİ İSTATİSTİĞİ ===")
if len(level_params_identified) > 0:
    constraint = " = 0, ".join(level_params_identified) + " = 0"
    wald_res = ardl_model.wald_test(constraint)
    print(wald_res)
else:
    print("Model parametreleri inceleniyor...")
    print(ardl_model.params)
```

```python
# Modeldeki düzey (level) değişkenlerinin tam parametre isimlerini tespit etme
# ARDL(1, 4, 1, 1) modelinde düzey terimleri: 'LOGGDP.L1', 'LOGINF.L0' (veya toplamı), 'LOGIMP.L0', 'LOGEXC.L0' vb.
print("Modeldeki tüm parametreler:")
print(ardl_model.params.index)

# Wald testi için kısıt oluşturma:
# Eşbütünleşme testi, gecikmeli bağımlı değişkenin (LOGGDP.L1) katsayısının 0'dan farklı olmasını
# ve bağımsız değişkenlerin düzey etkilerinin ortakça sıfıra eşitliğini sınar.
# Statsmodels ardl_model.wald_test() fonksiyonuna string kısıt verilir:
# Örnek: 'LOGGDP.L1 = 0'

constraints = ['LOGGDP.L1 = 0']
# Bağımsız değişkenlerin uzun dönem katsayıları veya düzey etkileri için kısıt ekleyebiliriz:
for param in ardl_model.params.index:
    if 'L' in param and param != 'LOGGDP.L1':
        # İsteğe bağlı olarak tüm gecikmeli terimler kısıota eklenebilir
        pass

wald_result = ardl_model.wald_test('LOGGDP.L1 = 0')
print("\n=== WALD TESTİ İSTATİSTİĞİ (EŞBÜTÜNLEŞME İÇİN) ===")
print(wald_result)
```

```python
# Katsayıları sözlük olarak alma
params = ardl_model.params

# LOGGDP.L1 katsayısı
gamma = params['LOGGDP.L1']
denom = 1 - gamma

# 1. Uzun Dönem Katsayıları Hesaplama
print("=== UZUN DÖNEM KATSAYILARI (MANUEL) ===")
# LOGINF uzun dönem katsayısı (LOGINF katsayıları toplamı / (1 - gamma))
loginf_sum = sum([params[p] for p in params.index if 'LOGINF' in p])
logimp_sum = sum([params[p] for p in params.index if 'LOGIMP' in p])
logexc_sum = sum([params[p] for p in params.index if 'LOGEXC' in p])

lr_inf = loginf_sum / denom
lr_imp = logimp_sum / denom
lr_exc = logexc_sum / denom

print(f"LOGINF (Enflasyon) Uzun Dönem Katsayısı: {lr_inf:.4f}")
print(f"LOGIMP (İthalat) Uzun Dönem Katsayısı: {lr_imp:.4f}")
print(f"LOGEXC (Döviz Kuru) Uzun Dönem Katsayısı: {lr_exc:.4f}")

# 2. Hata Düzeltme Katsayısı (ECT coefficient)
# ECM modelinde gecikmeli seviye bağımlı değişkenin katsayısı (gamma - 1) veya -(1 - gamma)
ect_coef = gamma - 1
print(f"\n=== HATA DÜZELTME KATSAYISI (ECT_t-1) ===")
print(f"Hata Düzeltme Hızı (ECT): {ect_coef:.4f}")
print(f"Yorum: Sistemdeki bir şokun / dengesizliğin %{abs(ect_coef)*100:.2f}'i bir sonraki dönemde (yılda/çeyrekte) bertaraf edilerek tekrar dengeye gelmektedir.")
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# Parametreler ve Varyans-Kovaryans Matrisi
params = ardl_model.params
cov = ardl_model.cov_params()
param_names = list(params.index)

gamma_idx = param_names.index('LOGGDP.L1')
gamma = params['LOGGDP.L1']
denom = 1 - gamma
df_resid = ardl_model.df_resid

def calculate_lr_with_delta(var_prefix):
    # İlgili değişkenin gecikme terimlerinin indekslerini bulma
    indices = [i for i, name in enumerate(param_names) if var_prefix in name and name != 'LOGGDP.L1']

    beta_sum = sum([params[i] for i in indices])
    lr_coef = beta_sum / denom

    # Delta Metodu için Gradyan (Türev) Vektörü
    grad = np.zeros(len(params))
    for i in indices:
        grad[i] = 1.0 / denom
    grad[gamma_idx] = beta_sum / (denom ** 2)

    # Varyans = grad^T * Cov * grad
    var_lr = grad.T @ cov @ grad
    se_lr = np.sqrt(var_lr)
    t_stat = lr_coef / se_lr
    p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df_resid))

    return lr_coef, se_lr, t_stat, p_val

variables = ['LOGINF', 'LOGIMP', 'LOGEXC']
results_list = []

for v in variables:
    lr, se, t_val, p_val = calculate_lr_with_delta(v)
    results_list.append({
        'Değişken': v,
        'Katsayı': round(lr, 4),
        'Std. Hata': round(se, 4),
        't-İstatistiği': round(t_val, 4),
        'p-değeri': round(p_val, 4)
    })

df_lr = pd.DataFrame(results_list)
print("=== DELTA METODU İLE UZUN DÖNEM KATSAYILARI VE İSTATİSTİKLERİ ===")
print(df_lr.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.ardl import ARDL
from scipy import stats

# 1. Veri setinin okunması ve logaritmik dönüşümler
df = pd.read_excel('Başarı.xlsx', sheet_name='EVDS')
for col in ['GDP', 'INF', 'IMP', 'EXC']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['LOGGDP'] = np.log(df['GDP'])
df['LOGINF'] = np.log(df['INF'])
df['LOGIMP'] = np.log(df['IMP'])
df['LOGEXC'] = np.log(df['EXC'])

# 2. ARDL(1, 4, 1, 1) modelinin kurulması ve tahmin edilmesi
model = ARDL(
    df['LOGGDP'],
    1,
    df[['LOGINF', 'LOGIMP', 'LOGEXC']],
    {'LOGINF': 4, 'LOGIMP': 1, 'LOGEXC': 1},
    trend='c'
)
results = model.fit()

# 3. Breusch-Godfrey Otokorelasyon Testi (Boyut uyumlu düzeltilmiş versiyon)
e = results.resid
n = len(e)
nlags = 2  # Test edilecek gecikme sayısı

# Efektif örneklem boyutuna göre exog matrisinin hizalanması
exog_eff = results.model.exog[-n:]
k_exog = exog_eff.shape[1]

# Tasarım matrisinin oluşturulması ve kalıntıların gecikmeleriyle yardımcı regresyon
Z = np.zeros((n - nlags, k_exog + nlags))
Z[:, :k_exog] = exog_eff[nlags:]
for i in range(nlags):
    Z[:, k_exog + i] = e[nlags - 1 - i : n - 1 - i]

e_trimmed = e[nlags:]
aux_ols = sm.OLS(e_trimmed, sm.add_constant(Z)).fit()
lm_stat = (n - nlags) * aux_ols.rsquared
p_val = 1 - stats.chi2.cdf(lm_stat, nlags)

# 4. Sonuçların Yazdırılması
print("=== BREUSCH-GODFREY OTOKORELASYON TESTİ ===")
print(f"LM Test İstatistiği: {lm_stat:.4f}")
print(f"p-değeri (p-value):   {p_val:.4f}")

if p_val < 0.05:
    print("Yorum: p < 0.05 olduğu için H0 reddedilir. Hata terimlerinde otokorelasyon vardır.")
else:
    print("Yorum: p > 0.05 olduğu için H0 reddedilemez. Hata terimlerinde otokorelasyon yoktur (İstenen durum).")
```

```python
from statsmodels.stats.diagnostic import het_breuschpagan, het_white

# Efektif kalıntılar ve tasarım matrisi
e = results.resid
n = len(e)
exog_eff = results.model.exog[-n:]
X_with_const = sm.add_constant(exog_eff)

# Breusch-Pagan Testi
bp_stat, bp_pval, bp_fstat, bp_fpval = het_breuschpagan(e, X_with_const)
print("=== BREUSCH-PAGAN TESTİ ===")
print(f"LM İstatistiği: {bp_stat:.4f}, p-değeri: {bp_pval:.4f}")

# White Testi
wh_stat, wh_pval, wh_fstat, wh_fpval = het_white(e, X_with_const)
print("\n=== WHITE TESTİ ===")
print(f"LM İstatistiği: {wh_stat:.4f}, p-değeri: {wh_pval:.4f}")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.ardl import ARDL
from statsmodels.stats.diagnostic import recursive_olsresiduals
import matplotlib.pyplot as plt

# 1. Veri setinin okunması ve logaritmik dönüşümler
df = pd.read_excel('Başarı.xlsx', sheet_name='EVDS')
for col in ['GDP', 'INF', 'IMP', 'EXC']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['LOGGDP'] = np.log(df['GDP'])
df['LOGINF'] = np.log(df['INF'])
df['LOGIMP'] = np.log(df['IMP'])
df['LOGEXC'] = np.log(df['EXC'])

# 2. ARDL(1, 4, 1, 1) modelinin kurulması
model = ARDL(
    df['LOGGDP'],
    1,
    df[['LOGINF', 'LOGIMP', 'LOGEXC']],
    {'LOGINF': 4, 'LOGIMP': 1, 'LOGEXC': 1},
    trend='c'
)
results = model.fit()

# 3. OLS eşdeğer modeli ve CUSUM / CUSUMSQ hesaplaması
y = results.model.endog
X = sm.add_constant(results.model.exog)
ols_res = sm.OLS(y, X).fit()

# Recursive kalıntılar ve CUSUM (%95 Güven Düzeyi)
recres = recursive_olsresiduals(ols_res, alpha=0.95)
skip = int(ols_res.df_model + 1)
rcusum = recres[5]
rcusumci = recres[6]

# Boyut hizalaması (CUSUM ve Sınırların Eşitlenmesi)
rcusum_plot = rcusum[len(rcusum) - len(rcusumci[0]):]
time_axis = np.arange(len(rcusum_plot))

# CUSUMSQ Hesaplaması
std_resids = recres[3][skip:]
cusumsq = np.cumsum(std_resids**2) / np.sum(std_resids**2)
nobs_eff = len(std_resids)
c_sq = 1.36 / np.sqrt(nobs_eff)

# 4. Grafiklerin Çizdirilmesi
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# CUSUM Grafiği
ax1.plot(time_axis, rcusum_plot, label='CUSUM', color='blue', lw=2)
ax1.plot(time_axis, rcusumci[0], label='%5 Kritik Sınır', color='red', linestyle='--')
ax1.plot(time_axis, rcusumci[1], color='red', linestyle='--')
ax1.set_title('CUSUM Kararlılık Testi', fontsize=12, fontweight='bold')
ax1.set_xlabel('Gözlem Dönemi')
ax1.legend()
ax1.grid(True, linestyle=':', alpha=0.6)

# CUSUMSQ Grafiği
ax2.plot(np.arange(len(cusumsq)), cusumsq, label='CUSUMSQ', color='darkgreen', lw=2)
ax2.plot(np.arange(len(cusumsq)), np.minimum(1.0, np.arange(1, len(cusumsq)+1)/len(cusumsq) + c_sq), color='red', linestyle='--', label='%5 Kritik Sınır')
ax2.plot(np.arange(len(cusumsq)), np.maximum(0.0, np.arange(1, len(cusumsq)+1)/len(cusumsq) - c_sq), color='red', linestyle='--')
ax2.set_title('CUSUMSQ Kararlılık Testi', fontsize=12, fontweight='bold')
ax2.set_xlabel('Gözlem Dönemi')
ax2.legend()
ax2.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
