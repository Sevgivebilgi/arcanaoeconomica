```python
import os
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt
```

```python
file_name = 'destan.xlsx.xlsx' if os.path.exists('destan.xlsx.xlsx') else 'destan.xlsx'
print(f"[BİLGİ] '{file_name}' dosyası okunuyor...")
df = pd.read_excel(file_name)
```

```python
df.columns = df.columns.astype(str).str.strip().str.upper()
```

```python
df[['LN_INFLATION', 'LN_IMPORT', 'LN_EXCHANGE', 'LN_GDP']] = np.log(df[['INFLATION', 'IMPORT', 'EXCHANGE', 'GDP']])
```

```python
vars_list = ['LN_INFLATION', 'LN_IMPORT', 'LN_EXCHANGE', 'LN_GDP']
col_mapping = {}
for target in vars_list:
    matched = [c for c in df.columns if target in c or target.replace('LN_', '') in c]
    if matched:
        col_mapping[matched[0]] = target

df = df.rename(columns=col_mapping)

print("\n" + "="*80)
print("1. BETİMSEL İSTATİSTİKLER (N=52)")
print("="*80)
print(df[vars_list].describe().T[['mean', 'std', 'min', 'max']])
```

```python
df = df[vars_list].copy()
```

```python
df = df.loc[:, ~df.columns.duplicated()]

print("\n" + "="*80)
print("1. BETİMSEL İSTATİSTİKLER (N=52)")
print("="*80)
print(df.describe().T[['mean', 'std', 'min', 'max']])
```

```python
print("\n" + "="*80)
print("2. ADF BİRİM KÖK TESTİ SONUÇLARI")
print("="*80)

adf_results = []
for var in vars_list:
    series_data = df[var].squeeze().dropna()

    # Seviye değerleri için sabitli (c), birinci farklar için sabitsiz/trendiz (n) veya sabitli test
    res_level = adfuller(series_data, autolag='AIC', regression='c')
    res_diff = adfuller(series_data.diff().dropna(), autolag='AIC', regression='c')

    p_lvl = res_level[1]
    p_diff = res_diff[1]

    # Entegrasyon derecesi kararı
    if p_lvl <= 0.05:
        order = 'I(0)'
    elif p_lvl > 0.05 and p_diff <= 0.05:
        order = 'I(1)'
    else:
        order = 'I(2) veya Durağan Değil'

    adf_results.append({
        'Değişken': var,
        'Seviye t-stat': round(res_level[0], 4),
        'Seviye p-val': round(p_lvl, 4),
        'Fark t-stat': round(res_diff[0], 4),
        'Fark p-val': round(p_diff, 4),
        'Entegrasyon Derecesi': order
    })

df_adf_summary = pd.DataFrame(adf_results)
print(df_adf_summary.to_string(index=False))
```

```python
df_ardl = pd.DataFrame({
    'Y': df['LN_INFLATION'].squeeze(),
    'Y1': df['LN_INFLATION'].squeeze().shift(1),
    'Y2': df['LN_INFLATION'].squeeze().shift(2),
    'X1': df['LN_IMPORT'].squeeze(),
    'X1_1': df['LN_IMPORT'].squeeze().shift(1),
    'X2': df['LN_EXCHANGE'].squeeze(),
    'X2_1': df['LN_EXCHANGE'].squeeze().shift(1),
    'X3': df['LN_GDP'].squeeze(),
    'X3_1': df['LN_GDP'].squeeze().shift(1)
}).dropna()

X = sm.add_constant(df_ardl.drop(columns=['Y']))
y = df_ardl['Y']

ardl_model = sm.OLS(y, X).fit()

print("\n" + "="*80)
print("3. ARDL MODEL TAHMİN ÖZETİ (OLS)")
print("="*80)
print(ardl_model.summary())
```

```python
print("\n" + "="*80)
print("4. PESARAN SINIR TESTİ (BOUNDS TEST)")
print("="*80)

f_test_bounds = ardl_model.f_test([
    'Y1 + Y2 = 1',
    'X1 + X1_1 = 0',
    'X2 + X2_1 = 0',
    'X3 + X3_1 = 0'
])

f_stat = float(f_test_bounds.fvalue)
print(f"Hesaplanan Wald F-İstatistiği: {f_stat:.4f}")
print("Pesaran (2001) k=3 Kritik Değerleri (%5 Düzeyi): I(0)=3.23 | I(1)=4.35")

if f_stat > 4.35:
    print(f"\nKARAR: F-istatistiği ({f_stat:.4f}) > 4.35 olduğu için H0 REDDEDİLDİ.")
    print("SONUÇ: Değişkenler arasında UZUN DÖNEMLİ EŞBÜTÜNLEŞME İLİŞKİSİ VARDIR.")
```

```python
print("\n" + "="*80)
print("5. UZUN DÖNEM ESNEKLİKLERİ VE ECM")
print("="*80)

params = ardl_model.params
adj_speed = 1 - params['Y1'] - params['Y2']

print(f"Uzun Dönem İthalat Fiyat Esnekliği (LN_IMPORT) : {((params['X1'] + params['X1_1']) / adj_speed):+.4f}")
print(f"Uzun Dönem Döviz Kuru Esnekliği  (LN_EXCHANGE): {((params['X2'] + params['X2_1']) / adj_speed):+.4f}")
print(f"Uzun Dönem Hasıla/Çıktı Esnekliği (LN_GDP)    : {((params['X3'] + params['X3_1']) / adj_speed):+.4f}")
print(f"Hata Düzeltme Katsayısı (ECM_t-1)             : {-adj_speed:.4f}")
```

```python
residuals = ardl_model.resid
cusum = np.cumsum(residuals) / np.std(residuals)
n = len(cusum)
critical_bound = 0.948 * np.sqrt(n)

fig, ax = plt.subplots(1, 2, figsize=(14, 4))

ax[0].plot(cusum, color='blue', label='CUSUM')
ax[0].axhline(critical_bound, color='red', linestyle='--')
ax[0].axhline(-critical_bound, color='red', linestyle='--')
ax[0].set_title('CUSUM Testi')
ax[0].grid(True, alpha=0.3)

cusum_sq = np.cumsum(residuals**2) / np.sum(residuals**2)
ax[1].plot(cusum_sq, color='green', label='CUSUMSQ')
ax[1].plot([0, n], [0, 1], color='red', linestyle='--')
ax[1].plot([0, n], [-0.1, 0.9], color='red', linestyle='--')
ax[1].set_title('CUSUMSQ Testi')
ax[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n[İŞLEM TAMAM] Tüm testler ve tablolar sorunsuz üretildi.")
```
